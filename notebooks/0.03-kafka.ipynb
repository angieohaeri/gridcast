# Kafka Streaming for live data (test)

```python
from confluent_kafka import Producer, Consumer
from confluent_kafka.admin import AdminClient, NewTopic
```

```python
broker = "angies-server:9094"
admin = AdminClient({"bootstrap.servers": broker})
```

```python
list(admin.list_topics(timeout=5).topics.keys())
```

```python
producer = Producer({"bootstrap.servers": broker})
producer.produce("test-topic", b"Hello, Kafka!")
producer.flush()
```

```python
consumer = Consumer({"bootstrap.servers": broker, "group.id": "test-group", "auto.offset.reset": "earliest"})
consumer.subscribe(["test-topic"])

msg = None
for _ in range(10):
    msg = consumer.poll(timeout=2)
    if msg is not None:
        break

print(f"Received message: {msg.value().decode('utf-8')}")
```
